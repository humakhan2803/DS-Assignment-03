```python
#An e-commerce store stores product detAn e-commerce store stores product details in a nested dictionary. Increase the Shirt stock from 200 to 250, then add a new product named Smartphone.
#ails in a nested dictionary. Increase the Shirt stock from 200 to 250, then add a new product named Smartp#hone.


products = {
    101: {"name": "Laptop", "category": "Electronics", "price": 1200, "stock": 50},
    102: {"name": "Shirt", "category": "Apparel", "price": 25, "stock": 200},
    103: {"name": "Coffee Maker", "category": "Home Appliances", "price": 80, "stock": 30}
}

# Increase the Shirt stock
products[102]["stock"] = 250
print("Updated inventory:")
print(products)

# Add a new product
products[104] = {
    "name": "Smartphone",
    "category": "Electronics",
    "price": 800,
    "stock": 100
}

print("\nInventory after adding Smartphone:")
print(products)
```

```python
#Ask the user for their age and classify them as Child (0–12), Teenager (13–19), Adult (20–64), or Senior (65+).
age = int(input("Enter your age: "))

if age < 0:
    print("Please enter a valid age")
elif age <= 12:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 64:
    print("Adult")
else:
    print("Senior")
```

```python
#Add Zara using the update() method and remove Omar using the pop() method.
student_details = {
    "Ali": {"age": 20, "grade": "A", "city": "Karachi"},
    "Sara": {"age": 21, "grade": "B", "city": "Lahore"},
    "Omar": {"age": 19, "grade": "A", "city": "Islamabad"},
    "Hina": {"age": 22, "grade": "C", "city": "Faisalabad"}
}

student_details.update({
    "Zara": {"age": 20, "grade": "B", "city": "Multan"}
})

student_details.pop("Omar")

print(student_details)
```

```python
#Add a new book using a dictionary method, display all book titles using a method, and remove books whose quantity is zero or less.
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

# Add a new book using update()
inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 4}
})

# Display all book titles using keys()
print("Book titles:")
print(list(inventory.keys()))

# Remove books that are out of stock
for title in list(inventory.keys()):
    if inventory[title]["quantity"] <= 0:
        inventory.pop(title)

print("\nUpdated inventory:")
print(inventory)
```

```python
#Ask for one character. Print Vowel if it is a vowel, Consonant if it is another alphabet letter, and a validation message otherwise.
ch = input("Enter one character: ")

if len(ch) != 1 or not ch.isalpha():
    print("Invalid input")
elif ch.lower() in "aeiou":
    print("Vowel")
else:
    print("Consonant")
```
